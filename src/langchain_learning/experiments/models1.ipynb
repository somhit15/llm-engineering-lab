```python
from langchain_learning.models import get_chat_model
from langchain.agents import create_agent

m1 = get_chat_model(max_tokens=2048, timeout=120)
```

```python
agent1 = create_agent(model=m1)
```

```python
mes1 = "You should be answering under 1000 tokens. How will you calculate the current temperature in New York City? and what all things you need for that"


res1 = agent1.invoke(
    {"messages":[{"role":"user", "content": mes1}]}
)

k = res1["messages"][-1].content
print(k)
```

```python
mes2 = "What should be a persons name if he works 12 hours regularly for 1 year"

agent2 = create_agent(model="google_genai:gemini-3.5-flash-lite",
                      system_prompt = "You should be answering in only 5 words, no more than that")

res2 = agent2.invoke(
    {"messages":[{"role":"user", "content": mes2}]}
)

print(res2["messages"][-1].content)
```
