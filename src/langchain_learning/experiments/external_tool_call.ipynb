```python
from tavily import TavilyClient
from langchain_learning.models import get_chat_model
from langchain.messages import AIMessage, HumanMessage, SystemMessage, ToolMessage
from langchain.tools import tool
from langchain.agents import create_agent, AgentState
from pydantic import BaseModel
from typing import Any
```

```python
m1 = get_chat_model(max_tokens = 30000, timeout = 120)
```

```python
tavily_client = TavilyClient()
```

```python
@tool
def web_search(query: str) -> dict[str, Any]:
    """ Search the web for information """
    return tavily_client.search(query)
```

```python
web_search.invoke("Who is the current chief minister of tamil nadu ?")
```

```python
a1 = create_agent(model = m1)

r1 = a1.invoke(
    {
        "messages": [HumanMessage("Who is the current chief minister of Tamil Nadu ?")]
    }
)

print(r1['messages'][-1].content)
```

```python
a2 = create_agent(model = m1, tools = [web_search])

r21 = a2.invoke(
    {
        "messages": [HumanMessage("Who is the current chief minister of Tamil Nadu ?")]
    }
)

print(r21['messages'][-1].content)
```
